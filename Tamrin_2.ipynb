# <div dir='rtl'>رسم نمودار Pie plot</div>

```python
df['Gender'].value_counts()
```

```python
df['Gender'].value_counts().plot(kind='pie');
```

# <div dir='rtl'>رسم نمودار Bar plot</div>

```python
df['Occupation'].value_counts().plot(kind='bar',colormap='coolwarm');
```

# <div dir='rtl'>رسم توزیع </div>

```python
df['Age'].plot(kind='kde');
```

# <div dir='rtl'>رسم Pair plot</div>

```python
sns.pairplot(df);
```

# <div dir='rtl'>رسم heatmap همبستگی</div>

```python
plt.figure(figsize=(12,8))
corr = df.iloc[:,[0,3,4,5,6,9,10,11,12,15,17,18]].corr()
sns.heatmap(corr,annot=True,cbar=True,fmt='.2f',cmap='coolwarm')
plt.title('Correlation Heatmap', fontsize=14);
```
